```python
import pandas as pd, seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')
df.info()         # struktur & tipe data
df.isnull().sum() # cek nilai hilang

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()        #matriks korelasi
```

```python
import pandas as pd, seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style='whitegrid')
df = sns.load_dataset('iris')
df.head()
```

langkah 2

```python
print(df.shape)
df.info()
df.describe()
```

langkah 3

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df['species'].value_counts())
```

langkah 4

```python
df.select_dtypes(include='number').hist(figsize=(9,6), bins=15)
plt.show()
sns.boxplot(data=df.select_dtypes(include='number'))
plt.title('deteksi outliter');
plt.show()
```

langkah 5

```python
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.show()
sns.pairplot(data=df, hue='species')
plt.show()
```

langkah 6

```python
num = df.select_dtypes(include='number')
sns.heatmap(num.corr(),annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```
