### Justifikasi Penggunaan `fit_transform` pada Data Latih dan `transform` pada Data Uji

Penerapan `fit_transform` pada *training data* dan `transform` secara eksklusif pada *test data* adalah praktik esensial dalam *data preprocessing* untuk menjaga validitas inferensi dan mencegah *data leakage*.

1.  **Metode `fit()`**: Mengestimasi parameter transformer (e.g., mean/std dev untuk `StandardScaler`, min/max untuk `MinMaxScaler`) secara eksklusif dari distribusi *training data* (`X_train`).
2.  **Metode `transform()`**: Menerapkan transformasi data menggunakan parameter-parameter yang telah diestimasi oleh `fit()`.
3.  **Metode `fit_transform()`**: Kombinasi `fit()` dan `transform()`, digunakan pada *training data* untuk estimasi parameter dan aplikasi transformasi secara bersamaan.

**Prinsip Utama:**

*   **Pencegahan *Data Leakage***: Estimasi parameter transformasi dari *test data* akan menginduksi *data leakage*, menghasilkan evaluasi model yang bias dan terlalu optimis. *Test data* harus tetap menjadi representasi data yang *unseen*.
*   **Konsistensi Transformasi**: Parameter transformasi yang diestimasi dari *training data* harus diterapkan secara konsisten pada *training data* maupun *test data*. Hal ini memastikan bahwa kedua set data berada dalam skala atau representasi fitur yang sama, menjaga validitas komparatif kinerja model.

Dengan demikian, `fit_transform` hanya digunakan pada data latih, sementara `transform` digunakan pada data uji dan data baru, untuk menjaga integritas statistik dan objektivitas evaluasi model *machine learning*.
